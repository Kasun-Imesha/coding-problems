## Linked List

a linear data structure, in which the elements are not stored at contiguous memory locations. The elements in a linked list are linked using pointers

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        

class LinekedList:
    def __init__(self):
        self.head = None
        
    def __repr__(self):
        # traverse the list and collect nodes' data
        nodes = []
        current = self.head
        while current:
            nodes.append(str(current.data))
            current = current.next
        return " -> ".join(nodes)
```

Your LinkedList class should support the following operations:

- int[] getValues() return an array of all the values in the linked list, ordered from head to tail.

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None
        
        # @property
        # def is_set(self):
        #     return self.val is not None
        
class LinkedList2:
    def __init__(self):
        self.head = None
        self.tail = None
        
    def get(self, index: int) -> int:
        cur = self.head
        i = 0
        while cur:
            # empty linked list
            if cur.val is None:
                return -1
            
            # matching index
            if i == index:
                return cur.val
            cur = cur.next
            i += 1
        
        # index out of bounds
        return -1
    
    def insert_head(self, val: int) -> None:
        new_head = Node(val)
        if self.head.val is None:
            self.head = new_head
            self.tail = new_head
        else:
            new_head.next = self.head
            self.head = new_head
            
    def insert_tail(self, val: int) -> None:
        new_tail = Node(val)
        if self.head.val is None:
            self.head = new_tail
            self.tail = new_tail
        else:
            self.tail.next = new_tail
            self.tail = new_tail
    
    def insert_at_index(self, index: int, val: int) -> None:
        assert index >= 0, "Index cannot be negative"
        
        if index == 0:
            self.insert_head(val)
            return 
        new_node = Node(val)
        cur = self.head
        i = 0
        
        while cur:
            if i == index - 1:
                new_node.next = cur.next
                cur.next = new_node
                
                if new_node.next is None:
                    self.tail = new_node
                return
            cur = cur.next
            i += 1
            
    def delete_at_index(self, index: int) -> None:
        assert index >= 0, "Index cannot be negative"
        
        if self.head.val is None:
            return
        
        if index == 0:
            self.head = self.head.next
        
        cur = self.head
        i = 0
        
        while cur:
            if i == index - 1:
                cur.next = cur.next.next
                if cur.next is None:
                    self.tail = cur
                return
            cur = cur.next
            i += 1  
            
    def get_values(self) -> list[int]:
        values = []
        cur = self.head
        while cur:
            values.append(cur.val)
            cur = cur.next
        return values
        
        
    
        
        
```

## Queue

a linear data structure that stores items in a First In First Out (FIFO) manner. With a queue, the least recently added item is removed first. A good example of the queue is any queue of consumers for a resource where the consumer that came first is served first.

- Enqueue: Adds an item to the queue. If the queue is full, then it is said to be an Overflow condition – Time Complexity: O(1)
- Dequeue: Removes an item from the queue. The items are popped in the same order in which they are pushed. If the queue is empty, then it is said to be an Underflow condition – Time Complexity: O(1)
- Front: Get the front item from queue – Time Complexity: O(1)
- Rear: Get the last item from queue – Time Complexity: O(1)

## Priority Queue

abstract data structures where each data/value in the queue has a certain priority. For example, In airlines, baggage with the title “Business” or “First-class” arrives earlier than the rest. Priority Queue is an extension of the queue with the following properties.

- An element with high priority is dequeued before an element with low priority.
- If two elements have the same priority, they are served according to their order in the queue.

### Custom Priority Queue Class

```python
class PriorityQueue(object):
    def __init__(self):
        self.deque = []
        
    def __str__(self):
        return "PriorityQueue(" + str(self.deque) + ")"
    def is_empty(self):
        return len(self.deque) == 0
    
    def insert(self, data):
        self.deque.append(data)
        # self.deque.sort()
        
    def remove(self):
        m = 0
        for i in range(len(self.deque)):
            if self.deque[i] > self.deque[m]:
                m = i
        
        ret = self.deque[m]
        del self.deque[m]
        
        return ret
        
```

```python
myQueue = PriorityQueue()
myQueue.insert(12)
myQueue.insert(1)
myQueue.insert(14)
myQueue.insert(7)
print(myQueue)            
while not myQueue.is_empty():
    print(myQueue.remove())
```

### Tree Traversal

#### Depth First Traversals

- Inorder (Left, Root, Right) : 4 2 5 1 3
- Preorder (Root, Left, Right) : 1 2 4 5 3
- Postorder (Left, Right, Root) : 4 5 2 3 1

**Algorithm Inorder(tree)**
- Traverse the left subtree, i.e., call Inorder(left-subtree)
- Visit the root.
- Traverse the right subtree, i.e., call Inorder(right-subtree)

**Algorithm Preorder(tree)**
- Visit the root.
- Traverse the left subtree, i.e., call Preorder(left-subtree)
- Traverse the right subtree, i.e., call Preorder(right-subtree)

**Algorithm Postorder(tree)**
- Traverse the left subtree, i.e., call Postorder(left-subtree)
- Traverse the right subtree, i.e., call Postorder(right-subtree)
- Visit the root.

**Time Complexity -  O(n)**

```python
# Binary tree
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
        
# inorder traversal
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)
        
# preorder traversal
def preorder(root):
    if root:
        print(root.data,  end=" ")
        preorder(root.left)
        preorder(root.right)
        
# postorder traversal
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")
```

```python
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.left = Node(6)
root.right.right = Node(7)
root.left.left.left = Node(8)
root.left.left.right = Node(9)
root.left.right.left = Node(10)
root.left.right.right = Node(11)

print("Inorder traversal: ", end="")
inorder(root)

print("\nPreorder traversal: ", end="")
preorder(root)

print("\nPostorder traversal: ", end="")
postorder(root)
```

### Adjacency Matrix Representation of a Graph

Adjacency Matrix is a 2D array of size V x V where V is the number of vertices in a graph. Let the 2D array be adj[][], a slot adj[i][j] = 1 indicates that there is an edge from vertex i to vertex j. The adjacency matrix for an undirected graph is always symmetric. Adjacency Matrix is also used to represent weighted graphs. If adj[i][j] = w, then there is an edge from vertex i to vertex j with weight w. 

```python
class Graph:
    def __init__(self, num_vertices):
        self.num_vertices = num_vertices
        self.vertices = {}
        self.vertices_list = [0] * num_vertices
        self.adj_matrix = [[-1] * num_vertices for _ in range(num_vertices)]
        
    def set_vertex(self, vertex, id):
        if 0<= vertex < self.num_vertices: 
            self.vertices[id] = vertex
            self.vertices_list[vertex] = id
        else:
            raise ValueError(f"Vertex out of bounds. Maximum allowed is {self.num_vertices}")
        
    def set_edge(self, from_vertex, to_vertex, cost=0):
        from_vertex = self.vertices[from_vertex]
        to_vertex = self.vertices[to_vertex]
        self.adj_matrix[from_vertex][to_vertex] = cost
        
        # for non directed graphs (for directed graphs do not include this line)
        self.adj_matrix[to_vertex][from_vertex] = cost
        
    def get_vertex(self):
        return self.vertices_list
    
    def get_edges(self):
        edges = []
        for i in range(self.num_vertices):
            for j in range(self.num_vertices):
                if self.adj_matrix[i][j] != -1:
                    edges.append((self.vertices_list[i], self.vertices_list[j], self.adj_matrix[i][j]))
                    
        return edges
    
    def get_matrix(self):
        return self.adj_matrix
```

```python
G = Graph(7)
G.set_vertex(0, "a")
G.set_vertex(2, "b")
G.set_vertex(3, "c")
G.set_vertex(4, "d")
G.set_vertex(5, "e")
G.set_vertex(6, "f")

G.set_edge("a", "e", 10)
G.set_edge("a", "c", 20)
G.set_edge("c", "b", 30)
G.set_edge("b", "e", 40)
G.set_edge("e", "d", 50)
G.set_edge("f", "e", 60)

print("Vertices of graph:")
print(G.get_vertex())

print("\nEdges of graph:")
print(G.get_edges())

print("\nAdjacency Matrix of graph:")
matrix = G.get_matrix()
for row in matrix:
    print(" ".join(map(str, row)))
```

### Adjacency List Representation of a Graph

An array of lists is used. The size of the array is equal to the number of vertices. Let the array be an array[]. An entry array[i] represents the **list of vertices \adjacent\ to the ith vertex**. This representation can also be used to represent a weighted graph. The weights of edges can be represented as lists of pairs.

graph[i] stores only ONE element - but that one element is the entry point (the "head") to a chain of linked nodes.

**What is a "Head"?**

The head is simply the first node in a linked list. graph[i] doesn't store a Python list - it stores a reference/pointer to the first AdjNode object.
Think of it like this:

- graph[0] is NOT a list
- graph[0] is a variable holding a reference to the first node
- That first node has a .next that points to the second node
- The second node has a .next that points to the third node
- And so on...

```python
# A class to represent the adjacency list of the node
class AdjNode:
    def __init__(self, data):
        self.vertex = data  # Which vertex we're connected to
        self.next = None  # Pointer to next connection (linked list)
        

# A class to represent a graph. A graph is the list of the adjacency lists.
# Size of the array will be the no. of the vertices "V"

class Graph:
    def __init__(self, vertices):
        self.V = vertices
        self.graph = [None] * self.V  # Array of V linked lists
    
    def add_edge(self, src, dest):
        # Step 1: Add dest to src's list
        node = AdjNode(dest)  # Create node for vertex dest
        node.next = self.graph[src]  # Point to current head of graph[src]
        self.graph[src] = node  # Make this the new head
        
        # Step 2: Add src to dest's list (undirected graph)
        node = AdjNode(src)  # Create node for vertex src
        node.next = self.graph[dest]  # Point to current head of graph[dest]
        self.graph[dest] = node  # Make this the new head
        
    def print_graph(self):
        for i in range(self.V):
            print(f"Adjacency list of vertex {i}\n head", end="")
            temp = self.graph[i]
            while temp:
                print(f" -> {temp.vertex}", end="")
                temp = temp.next
            print(" \n") 
    
```

```python
V = 5
graph = Graph(V)
graph.add_edge(0, 1)
graph.add_edge(0, 4)
graph.add_edge(1, 2)
graph.add_edge(1, 3)
graph.add_edge(1, 4)
graph.add_edge(2, 3)
graph.add_edge(3, 4)

graph.print_graph()
```

### Graph Traversal

#### Breadth-First Search or BFS

Breadth-First Traversal for a graph is similar to Breadth-First Traversal of a tree. The only catch here is, unlike trees, graphs may contain cycles, so we may come to the same node again. To avoid processing a node more than once, we use a boolean visited array. For simplicity, it is assumed that all vertices are reachable from the starting vertex.

For example, in the following graph, we start traversal from vertex 2. When we come to vertex 0, we look for all adjacent vertices of it. 2 is also an adjacent vertex of 0. If we don’t mark visited vertices, then 2 will be processed again and it will become a non-terminating process. A Breadth-First Traversal of the following graph is 2, 0, 3, 1.

```python
from collections import defaultdict
```

```python
class Graph:
    def __init__(self, is_directed=False):
        self.graph = defaultdict(list)
        self.is_directed = is_directed
        
    def add_edge(self, u, v):
        self.graph[u].append(v)
        
        if not self.is_directed:
            self.graph[v].append(u)  # For undirected graph
    
    def print_graph(self):
        for key in self.graph:
            print(f"{key} -> {' -> '.join(map(str, self.graph[key]))}")
            
    def BFS(self, start_vertex):
        visited = set()
        queue = []
        
        visited.add(start_vertex)
        queue.append(start_vertex)
        
        while queue:
            current = queue.pop(0)
            print(current, end=" ")
            
            for neighbor in self.graph[current]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
```

Time Complexity: O(V+E) where V is the number of vertices in the graph and E is the number of edges in the graph.

#### Depth First Search or DFS



Depth First Traversal for a graph is similar to Depth First Traversal of a tree. The only catch here is, unlike trees, graphs may contain cycles, a node may be visited twice. To avoid processing a node more than once, use a boolean visited array.

**Algorithm:**

- Create a recursive function that takes the index of the node and a visited array.
- Mark the current node as visited and print the node.
- Traverse all the adjacent and unmarked nodes and call the recursive function with the index of the adjacent node.

```python
from collections import defaultdict
```

```python
class Graph:

    # Constructor
    def __init__(self):

        # default dictionary to store graph
        self.graph = defaultdict(list)

    # function to add an edge to graph
    def add_edge(self, u, v):
        self.graph[u].append(v)

    # A function used by DFS
    def DFSUtil(self, v, visited):

        # Mark the current node as visited
        # and print it
        visited.add(v)
        print(v, end=' ')

        # Recur for all the vertices
        # adjacent to this vertex
        for neighbour in self.graph[v]:
            if neighbour not in visited:
                self.DFSUtil(neighbour, visited)

    # The function to do DFS traversal. It uses
    # recursive DFSUtil()
    def DFS(self, v):

        # Create a set to store visited vertices
        visited = set()

        # Call the recursive helper function
        # to print DFS traversal
        self.DFSUtil(v, visited)
```

```python
g = Graph()
g.add_edge(0, 1)
g.add_edge(0, 2)
g.add_edge(1, 2)
g.add_edge(2, 0)
g.add_edge(2, 3)
g.add_edge(3, 3)

print("Following is DFS from (starting from vertex 2)")
g.DFS(2)
```
