# Shortest Path Algorithms

## Bellman-Ford Optimization

The **Bellman-Ford algorithm** is a powerful tool for finding shortest paths from a single source. It is particularly robust because it can handle graphs with negative edge weights (unlike Dijkstra). The core concept is Relaxation: by checking all edges times, we propagate the shortest distance information across the entire network. Think of it like a water ripples expanding from the source; each iteration allows the "water" to travel one edge further to reach all possible nodes.

```
Approach:
- Distance Initialization: We set up a dist array of size . We set the source to 0 and all other potential destinations to Infinity.
- Iterative Relaxation: We run a loop times. In each pass, we look at every single edge in our list.
- If we have already found a path to (dist[u] != Infinity) and adding the weight gives us a better path to than what we currently have, we update dist[v].
- Maximum Delay Calculation: After the loops finish, the time it takes for the last node to receive the signal is simply the largest value in our dist array.
- Connectivity Check: If any node still has a distance of Infinity, it means that node is isolated from the source , so we return -1.
```

```
Complexity:
Time complexity: O(N * E) — We perform iterations, and in each iteration, we check all edges.
Space complexity: O(N) — We only store a single distance array of size .
```
