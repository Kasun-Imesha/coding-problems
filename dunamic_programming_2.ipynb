## 🔥 Step 1: What Dynamic Programming REALLY Is

DP is just:

Solving a big problem by solving smaller overlapping subproblems and storing their results.

Every DP problem has 2 properties:

Overlapping subproblems

Optimal substructure

If both exist → DP candidate.

## 🧠 Step 2: The 5-Step DP Thinking Framework

For EVERY DP problem, follow this checklist:

### 1️⃣ Define the State

Ask:

What does dp[i] (or dp[i][j]) represent?

This is the most important step.

Example:
```
dp[i] = best answer using first i elements

dp[i][j] = best answer using first i items with capacity j
```
If state is wrong → whole solution fails.

### 2️⃣ Write the Recurrence (Transition)

Ask:

If I know smaller answers, how do I compute the current one?

Usually:
```
dp[i] = min(...) or max(...) or sum(...)
```
### 3️⃣ Define Base Cases

What is the smallest valid subproblem?

### 4️⃣ Decide Computation Order

Top-down (recursion + memo)

Bottom-up (iterative)

In interviews → bottom-up is safer.

### 5️⃣ Optimize Space (if needed)

Many 2D DPs can be reduced to 1D.

## 📚 Step 3: Master These 7 Core Patterns

If you master these, 80% of interview DP is covered.

### 1️⃣ Fibonacci Pattern (1D DP)

Problems:

Fibonacci

Climbing stairs

House robber

Key idea:
```
dp[i] = dp[i-1] + dp[i-2]
```
### 2️⃣ 0/1 Knapsack Pattern

Problems:

Subset sum

Partition equal subset

Target sum

State:
```
dp[i][w] = best using first i items with capacity w
```
Transition:

Take item

Skip item

### 3️⃣ Unbounded Knapsack

Problems:

Coin change

Rod cutting

Difference:
You can reuse items.

### 4️⃣ Longest Increasing Subsequence (LIS)

State:
```
dp[i] = longest increasing subsequence ending at i
```
### 5️⃣ Longest Common Subsequence (LCS)

2D DP classic.

State:
```
dp[i][j] = LCS of first i chars of s1 and first j chars of s2
```
### 6️⃣ Interval DP

Problems:

Matrix chain multiplication

Burst balloons

State:

```
dp[i][j] = best answer in interval i to j
```
### 7️⃣ DP on Strings

Problems:

Edit distance

Palindrome partitioning

### 70. Climbing Stairs

https://leetcode.com/problems/climbing-stairs/description/

````python
class Solution:
    def climbStairs1(self, n: int) -> int:
        """
        time O(n)
        space O(n)
        """
        if n <= 2:
            return n

        dp = [0] * (n + 1)
        # base case
        dp[1] = 1
        dp[2] = 2

        for s in range(3, n + 1):
            # recurrence
            dp[s] = dp[s - 1] + dp[s - 2]

        return dp[n]

    def climbStairs(self, n: int) -> int:
        """
        we only use the previous two states to calculate the current state.
        time O(n)
        space O(1)
        """
        if n <= 2:
            return n

        prev1 = 1  # 1st stair
        prev2 = 2  # 2nd stair

        for s in range(3, n + 1):    
            curr = prev1 + prev2
            prev1, prev2 = prev2, curr
        
        return curr
````

### 198. House Robber

https://leetcode.com/problems/house-robber/description/

````python
class Solution:
    def rob1(self, nums: List[int]) -> int:
        """
        time: O(n)
        space: O(n)
        """
        n = len(nums)
        
        if n <= 2:
            return max(nums)
        
        dp = [0] * n
        # base cases
        dp[0] = nums[0]
        dp[1] = max(dp[0], nums[1])

        # recurrance
        for i in range(2, n):
            # state: pd[i] = maximum can be robbed from house 0 - i
            # we have two options at i, either rob (then we cannot rob i - 1) it or skip it (then we can rob i - 1).
            dp[i] = max(nums[i] + dp[i - 2], dp[i - 1])

        return dp[-1]

    def rob(self, nums: List[int]) -> int:
        """
        we can space optimize the above solution as we only need the previous two steps' cumulative amounts
        time: O(n)
        space: O(1)
        """
        n = len(nums)

        if n <= 2:
            return max(nums)

        # base cases
        rob0 = nums[0]
        rob1 = max(rob0, nums[1])

        for i in range(2, n):
            curr = max(rob0 + nums[i], rob1)
            rob0, rob1 = rob1, curr
        
        return curr
````

### 213. House Robber II

https://leetcode.com/problems/house-robber-ii/description/

````python
class Solution:
    
    """
    Either rob houses [0,1,2, ... n-2]  (exclude last)
    Or     rob houses [1,2,3, ... n-1]  (exclude first)
    
    One of these two will give the best answer. Why? Because in both cases you're guaranteed to never consider both house 0 and house n-1 together.
    """
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        if n <= 2:
            return max(nums)

        def sub_rob(sub_nums):
            sub_n = len(sub_nums)
            if sub_n <= 2:
                return max(sub_nums)

            rob0 = sub_nums[0]
            rob1 = max(sub_nums[0], sub_nums[1])

            for i in range(2, sub_n):
                curr = max(rob0 + sub_nums[i], rob1)
                rob0, rob1 = rob1, curr
            return curr
        
        return max(sub_rob(nums[1:]), sub_rob(nums[:n - 1]))
````

### Subset Sum Problem (416. Partition Equal Subset Sum)

https://leetcode.com/problems/partition-equal-subset-sum/description/

- Given an array nums and an integer target,
- return True if there exists a subset whose sum equals target

````python
class Solution:
    def canPartition1(self, nums: List[int]) -> bool:
        s = sum(nums)
        if s % 2 != 0:
            return False
        target = s // 2
        n = len(nums)

        # state
        # dp[i][s] - if sum `s` can be formed using numbers upto i
        dp = [[False] * (target + 1) for _ in range(n + 1)]

        # base cases
        # case 1: dp[i][0] = True -> without selecting anything we can form 0
        for i in range(n + 1):
            dp[i][0] = True

        # case 2: dp[0][s] = False; for s > 0 -> we cannot form positive sums without any elements
        for s in range(target + 1):
            dp[0][s] = False

        for i in range(1, n + 1):
            for s in range(1, target + 1):
                # recurrenece - skip or use nums[i]
                # skip nums[i]
                dp[i][s] = dp[i - 1][s]
                if s >= nums[i - 1]:
                    dp[i][s] = dp[i - 1][s] or dp[i - 1][s - nums[i - 1]]

        return dp[n][target]

    def canPartition(self, nums: List[int]) -> bool:
        """
        We only use previous row (i-1) so we can optimize the space to a 1D dp problem  
        """
        n = len(nums)
        if sum(nums) % 2 != 0:
            return False

        target = sum(nums) // 2
        dp = [False] * (target + 1)
        dp[0] = True  # sum 0 with empty set
        # state dp[s] means: using the elements processed so far, can we form sum s

        # we iterate backward in sums because each element can be used only once
        # If we iterate forward is s, the value dp[s - num] might have been updated in the same iteration using the same num.
        # So we allow, num + num + num + ... -> use the same element multiple times (unbounded knapsack problem)

        for num in nums:
            for s in range(target, num - 1, -1): # each element can be used only once
                dp[s] = dp[s] or dp[s - num]
        
        return dp[target]
````
