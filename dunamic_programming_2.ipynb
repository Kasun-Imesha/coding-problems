## 🔥 Step 1: What Dynamic Programming REALLY Is

DP is just:

Solving a big problem by solving smaller overlapping subproblems and storing their results.

Every DP problem has 2 properties:

Overlapping subproblems

Optimal substructure

If both exist → DP candidate.

## 🧠 Step 2: The 5-Step DP Thinking Framework

For EVERY DP problem, follow this checklist:

### 1️⃣ Define the State

Ask:

What does dp[i] (or dp[i][j]) represent?

This is the most important step.

Example:
```
dp[i] = best answer using first i elements

dp[i][j] = best answer using first i items with capacity j
```
If state is wrong → whole solution fails.

### 2️⃣ Write the Recurrence (Transition)

Ask:

If I know smaller answers, how do I compute the current one?

Usually:
```
dp[i] = min(...) or max(...) or sum(...)
```
### 3️⃣ Define Base Cases

What is the smallest valid subproblem?

### 4️⃣ Decide Computation Order

Top-down (recursion + memo)

Bottom-up (iterative)

In interviews → bottom-up is safer.

### 5️⃣ Optimize Space (if needed)

Many 2D DPs can be reduced to 1D.

## 📚 Step 3: Master These 7 Core Patterns

If you master these, 80% of interview DP is covered.

### 1️⃣ Fibonacci Pattern (1D DP)

Problems:

Fibonacci

Climbing stairs

House robber

Key idea:
```
dp[i] = dp[i-1] + dp[i-2]
```
### 2️⃣ 0/1 Knapsack Pattern

Problems:

Subset sum

Partition equal subset

Target sum

State:
```
dp[i][w] = best using first i items with capacity w
```
Transition:

Take item

Skip item

### 3️⃣ Unbounded Knapsack

Problems:

Coin change

Rod cutting

Difference:
You can reuse items.

### 4️⃣ Longest Increasing Subsequence (LIS)

State:
```
dp[i] = longest increasing subsequence ending at i
```
### 5️⃣ Longest Common Subsequence (LCS)

2D DP classic.

State:
```
dp[i][j] = LCS of first i chars of s1 and first j chars of s2
```
### 6️⃣ Interval DP

Problems:

Matrix chain multiplication

Burst balloons

State:

```
dp[i][j] = best answer in interval i to j
```
### 7️⃣ DP on Strings

Problems:

Edit distance

Palindrome partitioning

### 70. Climbing Stairs

https://leetcode.com/problems/climbing-stairs/description/

````python
class Solution:
    def climbStairs1(self, n: int) -> int:
        """
        time O(n)
        space O(n)
        """
        if n <= 2:
            return n

        dp = [0] * (n + 1)
        # base case
        dp[1] = 1
        dp[2] = 2

        for s in range(3, n + 1):
            # recurrence
            dp[s] = dp[s - 1] + dp[s - 2]

        return dp[n]

    def climbStairs(self, n: int) -> int:
        """
        we only use the previous two states to calculate the current state.
        time O(n)
        space O(1)
        """
        if n <= 2:
            return n

        prev1 = 1  # 1st stair
        prev2 = 2  # 2nd stair

        for s in range(3, n + 1):    
            curr = prev1 + prev2
            prev1, prev2 = prev2, curr
        
        return curr
````

### 198. House Robber

https://leetcode.com/problems/house-robber/description/

````python
class Solution:
    def rob1(self, nums: List[int]) -> int:
        """
        time: O(n)
        space: O(n)
        """
        n = len(nums)
        
        if n <= 2:
            return max(nums)
        
        dp = [0] * n
        # base cases
        dp[0] = nums[0]
        dp[1] = max(dp[0], nums[1])

        # recurrance
        for i in range(2, n):
            # state: pd[i] = maximum can be robbed from house 0 - i
            # we have two options at i, either rob (then we cannot rob i - 1) it or skip it (then we can rob i - 1).
            dp[i] = max(nums[i] + dp[i - 2], dp[i - 1])

        return dp[-1]

    def rob(self, nums: List[int]) -> int:
        """
        we can space optimize the above solution as we only need the previous two steps' cumulative amounts
        time: O(n)
        space: O(1)
        """
        n = len(nums)

        if n <= 2:
            return max(nums)

        # base cases
        rob0 = nums[0]
        rob1 = max(rob0, nums[1])

        for i in range(2, n):
            curr = max(rob0 + nums[i], rob1)
            rob0, rob1 = rob1, curr
        
        return curr
````

### 213. House Robber II

https://leetcode.com/problems/house-robber-ii/description/

````python
class Solution:
    
    """
    Either rob houses [0,1,2, ... n-2]  (exclude last)
    Or     rob houses [1,2,3, ... n-1]  (exclude first)
    
    One of these two will give the best answer. Why? Because in both cases you're guaranteed to never consider both house 0 and house n-1 together.
    """
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        if n <= 2:
            return max(nums)

        def sub_rob(sub_nums):
            sub_n = len(sub_nums)
            if sub_n <= 2:
                return max(sub_nums)

            rob0 = sub_nums[0]
            rob1 = max(sub_nums[0], sub_nums[1])

            for i in range(2, sub_n):
                curr = max(rob0 + sub_nums[i], rob1)
                rob0, rob1 = rob1, curr
            return curr
        
        return max(sub_rob(nums[1:]), sub_rob(nums[:n - 1]))
````

### Subset Sum Problem (416. Partition Equal Subset Sum)

https://leetcode.com/problems/partition-equal-subset-sum/description/

- Given an array nums and an integer target,
- return True if there exists a subset whose sum equals target

````python
class Solution:
    def canPartition1(self, nums: List[int]) -> bool:
        s = sum(nums)
        if s % 2 != 0:
            return False
        target = s // 2
        n = len(nums)

        # state
        # dp[i][s] - if sum `s` can be formed using numbers upto i
        dp = [[False] * (target + 1) for _ in range(n + 1)]

        # base cases
        # case 1: dp[i][0] = True -> without selecting anything we can form 0
        for i in range(n + 1):
            dp[i][0] = True

        # case 2: dp[0][s] = False; for s > 0 -> we cannot form positive sums without any elements
        for s in range(target + 1):
            dp[0][s] = False

        for i in range(1, n + 1):
            for s in range(1, target + 1):
                # recurrenece - skip or use nums[i]
                # skip nums[i]
                dp[i][s] = dp[i - 1][s]
                if s >= nums[i - 1]:
                    dp[i][s] = dp[i - 1][s] or dp[i - 1][s - nums[i - 1]]

        return dp[n][target]

    def canPartition(self, nums: List[int]) -> bool:
        """
        We only use previous row (i-1) so we can optimize the space to a 1D dp problem  
        """
        n = len(nums)
        if sum(nums) % 2 != 0:
            return False

        target = sum(nums) // 2
        dp = [False] * (target + 1)
        dp[0] = True  # sum 0 with empty set
        # state dp[s] means: using the elements processed so far, can we form sum s

        # we iterate backward in sums because each element can be used only once
        # If we iterate forward is s, the value dp[s - num] might have been updated in the same iteration using the same num.
        # So we allow, num + num + num + ... -> use the same element multiple times (unbounded knapsack problem) - try simulating with an example - don't mess up with repeted elements in the nums list

        for num in nums:
            for s in range(target, num - 1, -1): # each element can be used only once
                dp[s] = dp[s] or dp[s - num]
        
        return dp[target]
````

### Subset Sum Problem (494. Target Sum)

https://leetcode.com/problems/target-sum/description/

- Given an array nums and an integer target,
- return **the number of subsets** whose sum equals target

````python
class Solution:
    def findTargetSumWays(self, nums: List[int], target: int) -> int:
        """
        P - positive sum
        N - negative sum abs value
        P + N = total sum --(1)
        P - N = target    -- (2)
        (1) + (2) => 2P = total sum + target
        P = (total sum + target) / 2 => if we can find a positive sum (i.e. a subset with sum) (total sum + target) / 2,
        then it means, we can form the desired target using the aksed method
        """
        total_sum = sum(nums)
        n = len(nums)
        if (total_sum + target) % 2 != 0 or (total_sum + target) < 0: # the this is odd, then we cannot have an answer
            return 0
        
        psedo_target = (total_sum + target) // 2
        
        # state: dp[s] - how may subsets can sum up to s
        dp = [0] * (psedo_target + 1)
        dp[0] = 1  # one subset that sums to 0 - empty set

        # we iterate backward in sums because each element can be used only once
        # If we iterate forward is s, the value dp[s - num] might have been updated in the same iteration using the same num.
        # So we allow, num + num + num + ... -> use the same element multiple times (unbounded knapsack problem) - try simulating with an example  - don't mess up with repeted elements in the nums list
        for num in nums:
            for s in range(psedo_target, num - 1, -1): # each element can be used only once
                dp[s] += dp[s - num]
        
        return dp[psedo_target]
````

### 300. Longest Increasing Subsequence

https://leetcode.com/problems/longest-increasing-subsequence/description/

Given an array:
```
nums = [10, 9, 2, 5, 3, 7, 101, 18]
```
Find the length of the longest strictly increasing subsequence.

Example subsequence:
```
2 → 3 → 7 → 101
```
Length = 4

**Note:**
- A subsequence **does not need to be contiguous**.
- Order must be preserved.

````python
import bisect


class Solution:
    def lengthOfLIS1(self, nums: List[int]) -> int:
        """DP approach - O(n^2)"""
        n = len(nums)
        dp = [1] * n  # every index is a LIS of at least lenth 1
        # state dp[i] - LIS that ends at i

        for i in range(n):
            for j in range(i - 1, -1, -1):
                # state transition: for all j < i
                if nums[i] > nums[j]:
                    dp[i] = max(dp[i], 1 + dp[j])
        
        return max(dp)

    def bisect_right(arr, num):
        """arr must be a sorted one"""
        l = 0
        r = len(arr) - 1
        if num < arr[l]:
            return l
        if num >= arr[r]:
            return r + 1

        while l < r:
            mid = (l + r) // 2
            if num == arr[mid]:
                return mid + 1
            if num <= arr[mid]:
                r = mid + 1
            else:
                l = mid

    def bisect_left(arr, num):
        """arr must be a sorted one"""
        l = 0
        r = len(arr) - 1
        if num <= arr[l]:
            return l
        if num > arr[r]:
            return r + 1

        while l < r:
            mid = (l + r) // 2
            if num == arr[mid]:
                return mid
            if num < arr[mid]:
                r = mid + 1
            else:
                l = mid


    def lengthOfLIS(self, nums: List[int]) -> int:
        """non DP approach - O(nlog(n))"""    
        tail = [nums[0]]
        # tails[i] means:
        # the smallest possible ending value of an increasing subsequence of length i+1
        # tails[0] → smallest tail of subsequence length 1
        # tails[1] → smallest tail of subsequence length 2
        # tails[2] → smallest tail of subsequence length 3
        lis_len = len(tail)
        for num in nums[1:]:
            idx = bisect_left(tail, num)
            # idx = bisect_right(tail, num)
            # idx = bisect.bisect_left(tail, num)  # bisect_right gives the next index when a repeated number if found 
            # bisect_left([1,3,4], 3) = 1 -> [1, 3, 4]
            # bisect_right([1,3,4], 3) = 2 -> [1, 3, 3]
            # idx = bisect.bisect_right(tail, num)
            if idx < lis_len:
                tail[idx] = num
            else:
                tail.append(num)
                lis_len += 1

        print(tail)
        print(bisect_left([1,2,3,4], 3))
        print(bisect_right([1,2,3,4], 3))
        
        return lis_len
````

### 1143. Longest Common Subsequence

https://leetcode.com/problems/longest-common-subsequence/description/

````python
class Solution:
    def longestCommonSubsequence1(self, text1: str, text2: str) -> int:
            m = len(text1)
            n = len(text2)

            # state dp[i][j] - LCS of text1[0:i] and text2[0:j]
            # base bases dp[i][0] = 0 and dp[0][j] = 0 -> one string is empty -> 0 length LCS
            dp = [[0] * (n + 1) for _ in range(m + 1)]
            for i in range(m + 1):
                dp[i][0] = 0
            for j in range(n + 1):
                dp[0][j] = 0

            for i in range(1, m + 1):
                for j in range(1, n + 1):
                    # recurrance
                    # if a match
                    if text1[i - 1] == text2[j - 1]:
                        dp[i][j] = 1 + dp[i - 1][j - 1]
                    else: # no match
                        dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])
            
            return dp[m][n]

    def longestCommonSubsequence(self, text1: str, text2: str) -> int:
        m = len(text2)
        dp = [0]*(m+1)

        for i in range(1, len(text1)+1):
            prev_diag = 0

            for j in range(1, m+1):
                temp = dp[j]

                if text1[i-1] == text2[j-1]:
                    dp[j] = 1 + prev_diag
                else:
                    dp[j] = max(dp[j], dp[j-1])

                prev_diag = temp

        return dp[m]
````
