### 1. Subsets (LC 78)

```python
def subsets(nums):
    res = []
    def backtrack(start, path):
        res.append(path[:])
        for i in range(start, len(nums)):
            path.append(nums[i])
            backtrack(i + 1, path)
            path.pop()
    backtrack(0, [])
    return res
```

### 2. Subsets II (LC 90) — with duplicates

```python
def subsetsWithDup(nums):
    res = []
    nums.sort()
    def backtrack(start, path):
        res.append(path[:])
        for i in range(start, len(nums)):
            if i > start and nums[i] == nums[i - 1]:  # skip duplicates
                continue
            path.append(nums[i])
            backtrack(i + 1, path)
            path.pop()
    backtrack(0, [])
    return res
```

### 3. Permutations (LC 46)

```python
def permute(nums):
    res = []
    def backtrack(path, used):
        if len(path) == len(nums):
            res.append(path[:])
            return
        for i in range(len(nums)):
            if used[i]:
                continue
            used[i] = True
            path.append(nums[i])
            backtrack(path, used)
            path.pop()
            used[i] = False
    backtrack([], [False] * len(nums))
    return res
```

### 5. Combination Sum (LC 39) — reuse allowed

```python
def combinationSum(candidates, target):
    res = []
    def backtrack(start, path, remaining):
        if remaining == 0:
            res.append(path[:])
            return
        for i in range(start, len(candidates)):
            if candidates[i] > remaining:
                break
            path.append(candidates[i])
            backtrack(i, path, remaining - candidates[i])  # i, not i+1 (reuse allowed)
            path.pop()
    candidates.sort()
    backtrack(0, [], target)
    return res
```

### 6. Combination Sum II (LC 40) — no reuse, has duplicates

```python
def combinationSum2(candidates, target):
    res = []
    candidates.sort()
    def backtrack(start, path, remaining):
        if remaining == 0:
            res.append(path[:])
            return
        for i in range(start, len(candidates)):
            if candidates[i] > remaining:
                break
            if i > start and candidates[i] == candidates[i - 1]:  # skip duplicates
                continue
            path.append(candidates[i])
            backtrack(i + 1, path, remaining - candidates[i])
            path.pop()
    backtrack(0, [], target)
    return res
```

### 7. Combination Sum III (LC 216) — exactly k numbers summing to n, digits 1–9

```python
def combinationSum3(k, n):
    res = []
    def backtrack(start, path, remaining):
        if len(path) == k and remaining == 0:
            res.append(path[:])
            return
        if len(path) == k or remaining <= 0:
            return
        for i in range(start, 10):
            path.append(i)
            backtrack(i + 1, path, remaining - i)
            path.pop()
    backtrack(1, [], n)
    return res
```

### 8. Combination Sum IV (LC 377) — count of combinations (order matters → DP, not backtracking)

This problem asks for count where order matters, so it's best solved with DP. A pure backtracking TLEs on large inputs, but here's both:

```python
# DP solution (correct & efficient)
def combinationSum4(nums, target):
    dp = [0] * (target + 1)
    dp[0] = 1
    for t in range(1, target + 1):
        for num in nums:
            if t >= num:
                dp[t] += dp[t - num]
    return dp[target]

# Backtracking (correct but TLEs on large input)
def combinationSum4_backtrack(nums, target):
    count = [0]
    def backtrack(remaining):
        if remaining == 0:
            count[0] += 1
            return
        for num in nums:
            if num <= remaining:
                backtrack(remaining - num)
    backtrack(target)
    return count[0]
```

### 9. Palindrome Partitioning (LC 131)

```python
def partition(s):
    res = []
    def is_palindrome(sub):
        return sub == sub[::-1]
    def backtrack(start, path):
        if start == len(s):
            res.append(path[:])
            return
        for end in range(start + 1, len(s) + 1):
            substring = s[start:end]
            if is_palindrome(substring):
                path.append(substring)
                backtrack(end, path)
                path.pop()
    backtrack(0, [])
    return res
```

### 10. Generate Parentheses (LC 22)

```python
def generateParenthesis(n):
    res = []
    def backtrack(path, open_count, close_count):
        if len(path) == 2 * n:
            res.append("".join(path))
            return
        if open_count < n:
            path.append("(")
            backtrack(path, open_count + 1, close_count)
            path.pop()
        if close_count < open_count:
            path.append(")")
            backtrack(path, open_count, close_count + 1)
            path.pop()
    backtrack([], 0, 0)
    return res
```

### 11. N-Queens (LC 51)

```python
def solveNQueens(n):
    res = []
    cols = set()
    diag1 = set()  # row - col
    diag2 = set()  # row + col

    board = [["." ] * n for _ in range(n)]

    def backtrack(row):
        if row == n:
            res.append(["".join(r) for r in board])
            return
        for col in range(n):
            if col in cols or (row - col) in diag1 or (row + col) in diag2:
                continue
            cols.add(col)
            diag1.add(row - col)
            diag2.add(row + col)
            board[row][col] = "Q"
            backtrack(row + 1)
            cols.remove(col)
            diag1.remove(row - col)
            diag2.remove(row + col)
            board[row][col] = "."

    backtrack(0)
    return res
```
