## Find Square Root

```python
def interateSearch(n, lower_bound, upper_bound, th=0.0001):
    mid = (upper_bound + lower_bound) / 2
    while abs(n - mid * mid) > th:
        # print(n, lower_bound, upper_bound, mid)
        if n < mid * mid:
            upper_bound = mid
        else:
            lower_bound = mid
        
        mid = (upper_bound + lower_bound) / 2
    
    return mid
            
    
def findSqrt(n):
    if n < 0:
        return None
    
    # found = False
    
    i = 0
    while True:
        # print(i)
        if i * i == n:
            return i
        
        if i * i > n:
            print(n, i)
            return interateSearch(n, i - 1, i)

        i += 1
        
# findSqrt(0)
# findSqrt(1)
# findSqrt(100)
# findSqrt(2)
findSqrt(0), findSqrt(1), findSqrt(2), findSqrt(3), findSqrt(4), findSqrt(1000), findSqrt(1024)
```

## Any Root

```python
def interateSearch(n, root, lower_bound, upper_bound, th=0.0001):
    mid = (upper_bound + lower_bound) / 2
    while abs(n - mid ** root) > th:
        # print(n, lower_bound, upper_bound, mid)
        if n < mid ** root:
            upper_bound = mid
        else:
            lower_bound = mid
        
        mid = (upper_bound + lower_bound) / 2
    
    return mid
            
    
def findRoot(n, root=2):
    if n < 0:
        return None
    
    # found = False
    
    i = 0
    while True:
        # print(i)
        if i ** root == n:
            return i
        
        if i ** root > n:
            print(n, i)
            return interateSearch(n, root, i - 1, i)

        i += 1
        
# findRoot(0)
# findRoot(1)
# findRoot(100)
# findRoot(2)
findRoot(0), findRoot(1), findRoot(2), findRoot(3), findRoot(4), findRoot(1000), findRoot(1024, 4), findRoot(8, 3), findRoot(10, 5)
```
